```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
%matplotlib inline
```

```python
data = pd.read_excel(r"MBA_Dataset.xlsx")
data.head()
```

```python
basket = data.groupby(['Project_Name','month'])['Item_Name'].apply(list).reset_index()
basket
```

```python
from mlxtend.frequent_patterns import association_rules,apriori
from mlxtend.preprocessing import TransactionEncoder
```

```python
te = TransactionEncoder()
te_ary = te.fit(basket['Item_Name']).transform(basket['Item_Name'])
encoded_basket = pd.DataFrame(te_ary,columns = te.columns_)
encoded_basket
```

```python
freq_items = apriori(encoded_basket,min_support = 0.1,use_colnames = True)
```

```python
freq_items
```

```python
rules = association_rules(freq_items,metric = 'lift',min_threshold = 2)
rules
```

```python
rules.sort_values(by = 'lift',ascending = False,inplace = True)
```
