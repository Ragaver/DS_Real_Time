```python
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
import warnings
warnings.filterwarnings('ignore')
```

```python
data = pd.read_excel(r"C:\Users\91909\others\Downloads\mail_segregation.xlsx")
data.head()
```

```python
import re
```

```python
s1 = WordNetLemmatizer()
```

```python
corpus = []
for i in range(len(data)):
    messages = re.sub('[^a-zA-Z-0-9]',' ',data['Subject'][i])
    messages = messages.lower().split()
    messages = [s1.lemmatize(word,pos = 'v') for word in messages if word not in stopwords.words('english')]
    messages = ' '.join(messages)
    corpus.append(messages)
```

```python
v = CountVectorizer(max_features = 50,ngram_range = (1,3))
x = v.fit_transform(corpus)
x
```

```python
joblib.dump(v,'vectorizer.pkl')
```

```python
x =x.toarray()
```

```python
x
```

```python
x.shape
```

```python
y = data['Type']
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import BernoulliNB
```

```python
models = {'LogisticRegression':LogisticRegression(),
          'DecisionTreeClassifier':DecisionTreeClassifier(),
          'RandomForestClassifier':RandomForestClassifier(),
          'KNeighborsClassifier':KNeighborsClassifier(),
          'SVC':SVC(probability = True),
          'BernoulliNB':BernoulliNB()
         }
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,random_state = 10)
```

```python
ytest.shape
```

```python
for name,model in models.items():
    print(f'\n Model: {name}')
    mod = model.fit(xtrain,ytrain)
    pred = model.predict(xtest)
    print('Classification Report')
    print(f'{classification_report(pred,ytest)}')
    
    
```
